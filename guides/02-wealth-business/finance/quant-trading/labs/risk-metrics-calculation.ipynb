# Risk Metrics Calculation (VaR, Sharpe, Sortino)
- Compute metrics from PnL series


```python
import numpy as np
import pandas as pd

np.random.seed(0)
rets = np.random.normal(0.0005, 0.01, 1000)  # daily returns
ser = pd.Series(rets)

def metrics(r, freq=252, level=0.95):
    cagr = (1+r).prod()**(freq/len(r)) - 1
    vol = r.std()*np.sqrt(freq)
    sharpe = r.mean()*freq/(r.std()*np.sqrt(freq)) if vol!=0 else np.nan
    downside = r[r<0].std()*np.sqrt(freq)
    sortino = r.mean()*freq/downside if downside!=0 else np.nan
    var = np.percentile(r, (1-level)*100)
    cvar = r[r<=var].mean()
    curve = (1+r).cumprod()
    peak = curve.cummax()
    mdd = (curve/peak -1).min()
    return {'CAGR': cagr, 'Vol': vol, 'Sharpe': sharpe, 'Sortino': sortino, f'VaR_{int(level*100)}': var, f'CVaR_{int(level*100)}': cvar, 'MaxDD': mdd}

metrics(ser)
```
