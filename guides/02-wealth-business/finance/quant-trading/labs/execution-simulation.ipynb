# Execution Simulation (Slippage & Spread)
- Simple market order impact via bid-ask
- Add random slippage noise


```python
import numpy as np
import pandas as pd

np.random.seed(42)
n = 500
mid = 100 + np.cumsum(np.random.normal(0, 0.05, n))
spread = np.random.uniform(0.5, 1.5, n)/100  # 50-150 bps in cents
bid = mid - spread/2
ask = mid + spread/2
side = np.random.choice(['buy','sell'], size=n)
size = np.random.randint(10, 200, n)
slip = np.random.normal(0, 0.01, n)  # 1 cent std

fills = []
for i in range(n):
    px = ask[i] + slip[i] if side[i]=='buy' else bid[i] - slip[i]
    fills.append(px)

fills = np.array(fills)
impact = fills - mid
pd.Series(impact).describe()
```
