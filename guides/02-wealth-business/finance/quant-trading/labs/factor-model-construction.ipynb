# Multi-factor Model (Cross-sectional)
- Factors: size (ln mktcap), value (B/P), momentum (12-1m)
- Universe: sample from yfinance S&P500 tickers (subset)


```python
import yfinance as yf
import pandas as pd
import numpy as np

tickers = ['AAPL','MSFT','AMZN','GOOGL','META','JPM','XOM','CVX','HD','PG']
data = yf.download(tickers, start='2020-01-01', progress=False)
px = data['Adj Close'].dropna()
info = {t: yf.Ticker(t).info for t in tickers}
mktcap = pd.Series({t: info[t].get('marketCap', np.nan) for t in tickers})
book = pd.Series({t: info[t].get('bookValue', np.nan) for t in tickers})
price = px.iloc[-1]
bp = book * 1.0 / price
size = np.log(mktcap)
mom = (px.pct_change().add(1).cumprod().iloc[-22] / px.pct_change().add(1).cumprod().iloc[-252]) - 1 if len(px)>252 else pd.Series(np.nan, index=tickers)
factors = pd.DataFrame({'size': size, 'value_bp': bp, 'momentum_12m': mom})
factors.rank(ascending=False)
```
